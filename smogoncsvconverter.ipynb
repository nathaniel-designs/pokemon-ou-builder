```python
import re
import csv
import os
from pathlib import Path
```

```python
def txt_to_csv(input_file, output_file, delete_txt = False):
    input_path = Path(input_file)
    if not input_path.is_file():
        return 0 #This file is either invalid or has been converted already.
    else:
        pattern = re.compile(
            r'^\s*\|\s*(\d+)\s*\|\s*'
            r'(.+?)\s*\|\s*'
            r'([\d.]+)%\s*\|\s*'
            r'(\d+)\s*\|\s*'
            r'([\d.]+)%\s*\|\s*'
            r'(\d+)\s*\|\s*'
            r'([\d.]+)%\s*\|'
        )

        with open(input_file, "r", encoding="utf-8") as infile, \
            open(output_file, "w", newline="", encoding="utf-8") as outfile:

            writer = csv.writer(outfile)

            #Header
            writer.writerow([
                "Rank",
                "Pokemon",
                "USG%",
                "Raw",
                "Raw%",
                "Real",
                "Real%"
            ])

            for line in infile:
                match = pattern.match(line)
                if match:
                    writer.writerow(match.groups())

        print(f"Saved cleaned CSV to '{output_file}'")
        
        if delete_txt:
            os.remove(input_file)
            print(f"Deleted '{input_file}'")
```

```python
txt_to_csv("january_gen9ou_1500.txt", "january.csv", delete_txt = True)
txt_to_csv("february_gen9ou_1500.txt", "february.csv", delete_txt = True)
txt_to_csv("march_gen9ou_1500.txt", "march.csv", delete_txt = True)
txt_to_csv("april_gen9ou_1500.txt", "april.csv", delete_txt = True)
txt_to_csv("may_gen9ou_1500.txt", "may.csv", delete_txt = True)
txt_to_csv("june_gen9ou_1500.txt", "june.csv", delete_txt = True)
txt_to_csv("july_gen9ou_1500.txt", "july.csv", delete_txt = True)
txt_to_csv("august_gen9ou_1500.txt", "august.csv", delete_txt = True)
txt_to_csv("september_gen9ou_1500.txt", "september.csv", delete_txt = True)
```
